```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# --- Define the parameters for the normal distribution ---
mu = 0      # Mean for standard normal distribution
sigma = 1   # Standard deviation for standard normal distribution

# --- Define the test parameters ---
alpha = 0.05  # Significance level
z_critical = norm.ppf(1 - alpha)  # Z critical value for one-tailed test (right-tailed)
z_observed = -2.79 # The calculated Z-score from your analysis

# --- Define the range of Z values for the plot ---
z_values = np.linspace(-4, 4, 1000)
pdf = norm.pdf(z_values, mu, sigma) # Probability density function

# --- Plot the normal distribution curve ---
plt.figure(figsize=(10, 6))
plt.plot(z_values, pdf, label="Standard Normal Distribution", color='blue')

# --- Highlight the rejection region ---
plt.fill_between(z_values, 0, pdf, where=(z_values > z_critical), color='red', alpha=0.5, label=f'Rejection Region (α={alpha})')

# --- Annotate the Z critical value and the observed Z value ---
plt.axvline(z_critical, color='red', linestyle='--', label=f'Z Critical = {z_critical:.2f}')
plt.axvline(z_observed, color='black', linestyle='--', label=f'Observed Z = {z_observed}')

# --- Add labels, legend, and title ---
plt.title("One-Tailed Hypothesis Test (Right-Tailed)", fontsize=14)
plt.xlabel("Z-Score", fontsize=12)
plt.ylabel("Probability Density", fontsize=12)
plt.legend(fontsize=10)
plt.grid(alpha=0.3)

# --- Display the plot ---
plt.show()
```
